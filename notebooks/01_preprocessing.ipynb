```python
X = df.drop("default.payment.next.month", axis=1)
y = df["default.payment.next.month"]
```

```python
#creates synthetic samples of minority class using nearest neighbours
from imblearn.over_sampling import SMOTE

sm = SMOTE(random_state=42)
X_res, y_res = sm.fit_resample(X, y)
```

```python
print("Before SMOTE:", y.value_counts())
print("After SMOTE:", y_res.value_counts())
```

```python
processed_df = pd.concat([X_res, y_res], axis=1)
processed_df.to_csv(r"C:\Users\Chaitanya Sawant\OneDrive\Documents\Chaitanya\Projects\credit-defaulter\data\processed_data.csv", index=False)
```
