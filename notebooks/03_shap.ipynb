```python
import shap
import pandas as pd
import joblib
import matplotlib.pyplot as plt
```

```python
model = joblib.load('../models/rf_model.pkl')
scaler = joblib.load('../models/scaler.pkl')
```

```python
X = pd.read_csv('../data/processed_data.csv').drop(columns=["default.payment.next.month"])
y = pd.read_csv('../data/processed_data.csv')["default.payment.next.month"]
```

```python
X_scaled = scaler.transform(X)
```

```python
explainer = shap.Explainer(model, X_scaled) #auto chooses the right explainer type depending on the model
shap_values = explainer(X_scaled) #feature lvl contribution for each sample
```

```python
shap.summary_plot(shap_values, X, plot_type="bar")
```

```python
shap.summary_plot(shap_values, X)
```

```python
# Step 6: Local Explanation (Single Prediction)
# Choose one instance, e.g., index 5
sample_index = 5
shap.plots.force(explainer.expected_value, shap_values[sample_index], X.iloc[sample_index])
```
